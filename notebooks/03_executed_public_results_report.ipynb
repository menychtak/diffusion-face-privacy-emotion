# Public results report — NB24

**Executed summary notebook; not the original GPU generation/evaluation.** The underlying figures come from published aggregate results in `results/`. This notebook uses no face imagery or individual predictions. N23/N24 generation and evaluation were executed privately; their GitHub notebooks remain code-only.

The confidence interval below is **reported from the N24 paired bootstrap output**, not recalculated from the three aggregate means.

```python
from io import StringIO
import pandas as pd
# Frozen, source-independent NB24 means taken from results/nb24_heldout_aggregate_by_condition.csv
DATA = """condition,arcface_cosine,auraface_cosine,fer_matches_human_label,niqe
n07_g4p5,0.6556853239459021,0.6538824332809007,0.5740740740740741,7.701964658143655
n07_g5p5,0.4564701317016173,0.476257775530771,0.6203703703703703,7.612563207193323
same_expression_g4p5,0.5954521627820752,0.59030038566777,0.16666666666666666,7.8309388637908866
"""
result = pd.read_csv(StringIO(DATA)).set_index("condition")
print("NB24 final held-out — 108 source faces, 324 generated images")
print(result.to_string(float_format=lambda v: f"{v:.4f}"))
```

```python
contrast = result.loc["n07_g5p5"] - result.loc["n07_g4p5"]
print("Guidance 5.5 minus 4.5:")
print(f"ArcFace cosine: {contrast['arcface_cosine']:+.4f}")
print(f"AuraFace cosine: {contrast['auraface_cosine']:+.4f}")
print(f"FER accuracy: {100*contrast['fer_matches_human_label']:+.2f} percentage points")
print("Reported paired 95% bootstrap CI for FER difference: [-2.78, +12.04] pp")
print("Interpretation: FER improvement is not statistically established; cosine != re-identification rate.")
```
